### Section 5 - String methods

A **method** is a function that belongs to a value. You call it with a dot: `text.upper()`. `dir(...)` lists everything a string can do - it is a long list, so only the common ones matter now:

- `.upper()` / `.lower()` - change the case
- `.title()` - Capitalise Each Word
- `.strip()` - remove spaces at the start and end
- `.startswith(x)` / `.endswith(x)` - True or False
- `.replace(old, new)` - swap one piece of text for another
